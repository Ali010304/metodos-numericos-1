```python
import numpy as np
import matplotlib.pyplot as plt
def f(x):
  return np.exp(x**2)

xx = np.linspace(-1.1, 1.1, 100)
plt.plot(xx, f(xx))
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Plot of f(x) = exp(x^2)")
plt.grid(True)
plt.show()
```
