Método de Bisección

  Imaginemos un juego, nos piden adivinar un numero entre otros dos números. Es decir hallar un $p \in [a,b]$.
  Si nuestro número

Se desea resolver la ecuacion $x^3-4x^2-10=0$ en el intervalo $[1,2]$.

Este método evalúa cada extremo en $f(x)$ para verificar que tenga signos opuestos, entonces por el teorema del valor intbrmedio, este procedimiento asegura que existe un $p \in [a,b]$: $p$ es raíz de $f(x)$, es decir, $f(p)=0$

La aproximacion de esta raíz se hace mediante la siguinete fórmula:
$$p_i= \frac{b_i - a_i}{2}$$

```python
from math import *
import numpy as np
import matplotlib . pyplot as plt
```

Librerias que permiten realizar funciones matematicas (math), matplotlib permite graficar y numpy el manejo de arreglos numericos.

```python
def f(x):
  return x**3 +4*x**2 - 10.0
```

Se define la función
en la que se trabajara.

```python
x=np.linspace(0,3,50)
plt.plot(x,f(x))
plt.title("Grafica $f(x)=x^3+4x^2-10$")
plt.axhline (y=0, color='red')
```

Se muestra una gráfica de la función, la recta de color rojo muestra el punto exacto donde la corta.  
Este punto en particular es la raíz de la función.

```python
#1
a=1.0
b=2.0

#2
a0=a
b0=b

#3
tol=0.00001
nmax=100
error=100
niter=0
```

La parte #1 establece el intervalo donde se desea hallar la raíz de la función.

Debajo, #2 se encarga de guardar una copia de los valores anteriores, ya que conforme el método avanza estos mismos se irán modificando. Con cada iteración  el intervalo se hará más pequeño.

Por último, #3 ahí se encargan de limitar la longitud del intervalo y el número de iteraciones asi como el mismo bucle.



```python

# Método de Bisección
m = a + (b - a)/2

#Evaluacion de la función en los puntos a, b y m
fa = f(a)
fb = f(b)
fm = f(m)

print("# iter\t\t a \t\t f(a) \t\t b \t\t f(b) \t\t m \t\t f(m) \t\t error")
print("{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter, a0, fa, b0, fb, m, fm, error ))
```

En este apartado se comienza el cálculo y proceso de evaluación del punto medio, mediante la fórmula:

$$p_i= \frac{b_i - a_i}{2}$$

El resultado de esta operación dará un número que será evaluado posteriormente en la ecuación dada, para finalmente ser impreso en pantalla.

Esto da inicio a las iteraciones, siendo esesta la iteracion número 0.

```python

# ciclo iterativo
while error > tol and niter < nmax:
    m = a + (b - a) / 2
    if np.sign(fa) == np.sign(fm):
        a = m
        fa = f(a)
    else:
        b = m
        fb = f(b)

    m = a + (b - a)/2
    fm = f(m)
    error = abs(b - a)
    niter += 1
    print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f} \t {4:6.6f} \t {5:6.6f} \t {6:6.6f} \t {7:6.6f}".format(niter, a, fa, b, fb, m, fm, error ))

print("La raíz de la función dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.7f}".format(a0,b0,m))
```

En este último bloque se halla el ciclo que controla las iteraciones y que se encarga de encontrar la raíz.

Dentro del ciclo se comparan signos, si $f(a)$ y $f(b)$ tienen el mismo signo entonces se descarta la mitad de la parte izquierda, si no es así, entonces se descarta la mitad de la derecha.

Entonces, dependiendo del signo será la forma en la que se acorta el nuevo intervalo hasta irse reduciendo cada vez más.

Se usan condicionales precisamente para vverificar que lado es el que se usará, además el error permite conocer que tanta distancia hay entre los nuevos intervalos.

Por último, se anota cada iteracion en una tabla permitiendo mostrar los resutados de cada itnento.
