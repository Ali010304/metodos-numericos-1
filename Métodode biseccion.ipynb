# **Método de Bisección**

  Imaginemos un juego, se nos pide adivinar un número escondido entre otros dos números, es decir; encontrar un $p \in [a,b]$.

  Dividimos a la mitad el intervalo, si nuestro número resulta estar más cerca de $b$, entonces descartamos la parte izquierda, que va desde $a$ hasta $p$. De manera análoga, si resulta  ser más cercano a $a$ entonces se elimina la parte derecha, que va desde $p$ hasta $b$.

  A continuación, volvemos a dividir a la mitad el nuevo intervalo que será $[a, p]$ o $[p, b]$, según el resultado obtenido. Si repetimos este proceso de cortar siempre a la mitad, llegará un momento en el que nos aproximaremos tanto al número que obtendremos la respuesta con la precisión deseada.

Este es el método de bisección de forma intuitiva,pero resulta ser un poco más fácil de entender si se explica mediante fórmulas matemáticas.


Sea $f(x)$, una función continua en el intervalo $[a,b]$: $f(a)$ y $f(b)$ tienen signos opuestos ($f(a)* f(b)$<$0$) entonces por el Teorema del Valor Intermedio (T.V.I.) $\exists$ $p \in (a,b): f(p)=0$



## s.png



Ejemplo:

Se desea resolver la ecuacion $x^3+4x^2-10=0$ en el intervalo $[1,2]$.
Primero evaluamos $f(a)$ y $f(b)$.

 $f(1)= (1)^3+4(1)^2-10=-5 < 0$.

 $f(2)= (2)^3+4(2)^2-10=14 > 0$.

Como ambas evaluaciones tienen signos opuestos, el T.V.I garantiza la existencia de un $p \in [a,b]: f(p)=0$, es decir; la raíz de la función.

La aproximacion de esta raíz se hace mediante la siguiente fórmula:
$$p_i= \frac{a_i + b_i}{2}$$

Entonces
$$p_1= \frac{1+2}{2}$= $1.5$$

Evaluamos $f(1.5)$ = $2.375 > 0$, esto nos indica que debemos usar el intervalo $[1, 1.5]$

 $$ p_2= \frac{1+1.5}{2} = 1.25$$

Evaluamos $f(1.25)$ = $-1.796875$, nos indica el nuevo intervalo $[1.25,1.5]$

Si se continua de esta manera, a cierto número de iteraciones se llegará a un valor aproximado, a continuacion se probara una mejor manera de obtenerlo.
Después de la iteración número 13, $p = 1.365234375$ siendo esta la aproximación más exacta de la raíz.

Aunque este método es intuitivo tiene desventajas, una de ellas es que su velocidad para calcular la exactitud suele ser lenta, pues requiere de un gran número de iteraciones para conocer la aproximación más exacta. Es por ello que a continuación se presentará una forma más rápida.

```python
#1
a=1.0
b=2.0

#2
a0=a
b0=b

#3
tol=0.00001
nmax=100
error=100
niter=0
```

La parte #1 establece el intervalo donde se desea hallar la raíz de la función.

Debajo, #2 se encarga de guardar una copia de los valores anteriores, ya que conforme el método avanza estos mismos se irán modificando. Con cada iteración  el intervalo se hará más pequeño.

Por último, #3 ahí se encargan de limitar la longitud del intervalo y el número de iteraciones asi como el mismo bucle.



```python

# Método de Bisección
m = a + (b - a)/2

#Evaluacion de la función en los puntos a, b y m
fa = f(a)
fb = f(b)
fm = f(m)

print("# iter\t\t a \t\t f(a) \t\t b \t\t f(b) \t\t m \t\t f(m) \t\t error")
print("{0} \t\t {1:6.4f} \t {2:6.4f} \t {3:6.4f} \t {4:6.4f} \t {5:6.4f} \t {6:6.4f} \t {7:6.4f}".format(niter, a0, fa, b0, fb, m, fm, error ))
```

En este apartado se comienza el cálculo y proceso de evaluación del punto medio.

El resultado de esta operación dará un número que será evaluado posteriormente en la ecuación dada, para finalmente ser impreso en pantalla.

Esto da inicio a las iteraciones, siendo esta la iteracion número 0.

```python

# ciclo iterativo
while error > tol and niter < nmax:
    m = a + (b - a) / 2
    if np.sign(fa) == np.sign(fm):
        a = m
        fa = f(a)
    else:
        b = m
        fb = f(b)

    m = a + (b - a)/2
    fm = f(m)
    error = abs(b - a)
    niter += 1
    print("{0} \t\t {1:6.6f} \t {2:6.6f} \t {3:6.6f} \t {4:6.6f} \t {5:6.6f} \t {6:6.6f} \t {7:6.6f}".format(niter, a, fa, b, fb, m, fm, error ))

print("La raíz de la función dada en el intervalo [{0:6.4f},{1:6.4f}] es {2:6.7f}".format(a0,b0,m))
```

En este último bloque se halla el ciclo que controla las iteraciones y que se encarga de encontrar la raíz.

Dentro del ciclo se comparan signos, si $f(a)$ y $f(m)$ tienen el mismo signo entonces se descarta la mitad de la parte izquierda, si no es así, entonces se descarta la mitad de la derecha.

Dependiendo del signo será la forma en la que se acorta el nuevo intervalo hasta irse reduciendo cada vez más.

Se usan condicionales precisamente para verificar que lado es el que se usará, además el error mide qué tan ancho es el rango donde todavía está escondida la raíz. Mientras más pequeño sea ese rango, más se reduce el margen de equivocación y más cerca está el resultado exacto.

Por último, se anota cada iteracion en una tabla permitiendo mostrar los resutados de cada intento.

Finalmente se nos entrega el resultado.
La raíz de la función $x^3+4x^2-10=0$ en el intervalo $[1,2]$ es **1.3652306**
