# **Método del punto fijo**
Imaginemos un entrenamiento de tiro de arco. En el entrenamiento de hoy se nos pide poder disparar al centro de la diana, la cuestion es la siguinte:


*   En el primer tiro fallamos y la flecha pasa de largo del objetivo, nos movemos un poco de nuestro lugar y preparamos el tiro.
*   El segundo tiro es mejor que el primero, porque aunque no dió en el blanco esta vez se clavo en las orillas de la diana, nos movemos un poco más y preparamos el disparo.
* El tercer tiro mejora considerablemente con respecto al segundo, y es que esta vez nos acercamos más al centro de la diana, nos movemos ligeramente y disparamos.
* El cuarto tiro se acerca cada vez más y el quinto aún más.

Repitiendo este mismo proceso, cada intento hecho será el punto de partida para el siguiente intento. Después de varios intentos podremos dar justo en el centro de la diana.

A ese lugar lo conoceremos como **punto fijo** y es aquel lugar al que logramos llegar tras repetir un proceso varias veces, y que una vez alcanzado hace que nuestro resultado  no cambie.





De una manera más formal, podemos definir al **punto fijo** de la siguiente manera:

Sea $g \in C[a,b]$, se dice que **$p \in [a,b]$** es un **punto fijo** de $g$ si se satisface que:

$$g(p) = p$$

Además:

1. **Existencia:** Si se cumple que:

   $$g(x) \in [a,b] \quad \forall x \in [a,b]$$

   entonces la función $g$ tiene **al menos un punto fijo** $p \in [a,b]$.

2. **Unicidad:** Si además existe la derivada $g'(x)$ en $(a,b)$ y existe una constante $k$ tal que $0 < k < 1$ que satisface:

   $$|g'(x)| \le k \quad \forall x \in (a,b)$$

   entonces existe **exactamente un único punto fijo** $p \in [a,b]$.

3. **Error:** Si $p$ es el punto fijo y $p_n$ la aproximación obtenida en la iteración $n$, el error es:

$$
|e|=|p-p_n|
$$

Como normalmente no conocemos $p$, utilizamos como criterio la diferencia entre dos aproximaciones consecutivas:

$$|e|= |x_{n+1}-x_n|<\varepsilon $$

donde $\varepsilon$ es la tolerancia establecida.

Es decir, al realizar una iteracion restamos el resultado de la evaluación  y el punto donde se evaluó.

# Ejemplo



Hallar el punto fijo de la ecuación

$$f(x) = x^3+4x^2-10=0$$

Despejamos la ecuación  y obtenemos:


*   $x=g_1(x)= \sqrt{10-4x^2}$

*   $x=g_2(x)$ = $\frac{1}{2}$ $\sqrt{10-x^3}$

Ahora bien, una duda recurrente es el decidir que ecuación utilizar, para ello aplicamos la condicion de unicidad, entonces derivamos cada ecuacion y evaluamos en un punto inicial, en este caso $p_0= 1.5$



*   $g_1´(x) = \frac {-4x}{\sqrt {10-4x^2}}$ $\implies$ $|g_1´(1.5)| = |\frac{-6}{\sqrt{1}}| = |-6| = 6 $

**No se cumple**

$$|g'(x)| \le 1$$

$$|g_1´(1.5)| > 1$$

$\therefore g_1(x)$ no sirve ya que diverge al usar este método.




*   $g_2´(x) = \frac {-3x^2}{4 \sqrt {10-x^3}}$ $\implies$ $|g_2´(1.5)| = |\frac{-6.75}{4 \sqrt{6.625}}| \approx |-.655| = .655 $

**Cumple con  la condición**

$$|g'(x)| \le 1$$

$$|g_2´(1.5)| < 1$$

$\therefore g_2´(x)$ nos sirve para usar este método





Entonces, tomando como punto inicial $p_0 = 1.5$, y al evaluar en $g_2(x)$ = $\frac{1}{2}$ $\sqrt{10-x^3}$ genera los siguientes resultados:

* **Iteración 1:**
   * $p_1 = g(1.5) = \frac{1}{2}\sqrt{10 - (1.5)^3} = \mathbf{1.2869537}$
   * $\text{Error} = |1.2869537 - 1.5| = \mathbf{0.2130463}$

* **Iteración 2:**
   * $p_2 = g(1.2869537) = \frac{1}{2}\sqrt{10 - (1.2869537)^3} = \mathbf{1.4025408}$
   * $\text{Error} = |1.4025408 - 1.2869537| = \mathbf{0.1155871}$

* **Iteración 3:**
   * $p_3 = g(1.4025408) = \frac{1}{2}\sqrt{10 - (1.4025408)^3} = \mathbf{1.3454583}$
   * $\text{Error} = |1.3454583 - 1.4025408| = \mathbf{0.0570825}$

* **Iteración 4:**
   * $p_4 = g(1.3454583) = \frac{1}{2}\sqrt{10 - (1.3454583)^3} = \mathbf{1.3751702}$
   * $\text{Error} = |1.3751702 - 1.3454583| = \mathbf{0.0297119}$

* **Iteración 5:**
   * $p_5 = g(1.3751702) = \frac{1}{2}\sqrt{10 - (1.3751702)^3} = \mathbf{1.3600941}$
   * $\text{Error} = |1.3600941 - 1.3751702| = \mathbf{0.0150761}$

* **Iteración 6:**
   * $p_6 = g(1.3600941) = \frac{1}{2}\sqrt{10 - (1.3600941)^3} = \mathbf{1.3678183}$
   * $\text{Error} = |1.3678183 - 1.3600941| = \mathbf{0.0077242}$

   Como se puede observar por cada interación que se realiza el resultado se va dirigiendo a una raíz de la ecuación, mientras que el error nos va diciendo que tantanta exactitud estamos teniendo.

   En este caso particular, la raíz  se puede observar al realizar apenas seis iteraciones, pero esto no siempre será así, es por ello que tenemos una forma más rapida de lograrlo sin tener que realizar iteraciones infinitas manualmente.

## Codigo para encontrar el punto fijo de una ecuación



```python

from math import *
import numpy as np
import matplotlib . pyplot as plt
```

Librerias necesarias para calculos matematicos, realizar graficas y parael control de iteraciones.


```python
def f(x):
  return x**3 +4*x**2 - 10.0
```

Se define la función con la que se ha estado trabajando

```python
a = 1.0
b = 2.0
p0 = 1.5
raiz_aprox = 1.365

x=np.linspace(0,3,50)
plt.plot(x,f(x))
plt.title("Grafica $f(x)=x^3+4x^2-10$")
plt.axhline (y=0, color='red')
plt.axvline(x=a, color="orange", linestyle=":", label="Intervalo $[a, b] = [1, 2]$")
plt.axvline(x=b, color="orange", linestyle=":")
plt.axvline(x=p0, color="purple", linestyle="--", label="Punto medio $p_1 = 1.5$")
plt.xlabel("$x$")
plt.ylabel("$f(x)$")
plt.plot(raiz_aprox, 0, "go", markersize=8, label="Raíz $r \approx 1.365$")
```

Al ser la misma función que en el Método de Bisección, la gráfica es la misma y los elementos también lo son.

- La recta de color rojo muestra el punto ddonde corta a la función y el lugar exacto donde se encuentra la raíz.

- La línea punteada color morada muestra el punto de partida $p_0$ donde se iniciará la evaluación de $g_2(x)$.

- Las dos lineas punteadas color naranjas muestran el intervalo inicial $[1,2]$.

- El punto verde entre todas las líneas muestran la el lugar de la raíz exacta.

```python
def g(x):
    return 0.5 * np.sqrt(10 - x**3)

pfijo(g, p0=0.5, n_iter=25)

def pfijo(g, p0, n_iter):

    print("{:<5} {:<15} {:<15} {:<15}".format("I", "P_0", "P_n  = g(p0)", "Error = |p1-p0|"))
    print("-" * 55)

    for i in range(1, n_iter + 1):
        p1 = g(p0)
        e = np.abs(p1 - p0)
        print("{:<5d} {:<15.7f} {:<15.7f} {:<15.7f}".format(i, p0, p1, e))

        p0 = p1

    print("-" * 55)

    print("La raíz de la función después de n iteraciones en el intervalo [1,2] es 1.365230")
```

El responsable de todo el trabajo es este pedazo, dentro de esta celda se realizan las operaciones y las iteraciones necesarias para poder aproximarnos a la raíz.
Utilizamos el ciclo "for" para controlar el número de iteraciones que se desean realizar, se calcula el error ya que este nos indicara que tanta precisión con el resultado tenemos

Como podemos observar, el resultado al que se llega resulta ser el mismo que obtuvimos en el Método de Biseccion, ambos métodos cumplen la misma función de poder encontrar la raíz de una ecuación en un intervalo específico, la elección entre un método y otro dependerá de la manera en la que busquemos trabajar y de la cantidad de tiempo que nos queramos tomar.
